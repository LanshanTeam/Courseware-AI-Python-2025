# LangGraph

LangGraph是一个为 AI Agent 应用程序提供持久运行时的框架。是一个低级编排框架和运行时，用于构建、管理和部署长期运行的有状态 Agent。

LangGraph专注于 Agent 编排所需的底层能力：持久执行、流式传输、人机参与循环等。

官方文档：https://docs.langchain.com/oss/python/langgraph/overview

## LangGraph Orientation

### 为什么要使用 LangGraph？

- Why：构建AI应用程序很容易入门，但很难定制和扩展。
- 目标：让 LangGraph 成为在生产环境中运行代理的框架

### 使用 LLM 编写应用程序所面临的挑战

- **延迟**
    - LLM 的延迟非常长，有时以秒为单位，而不是毫秒。当前的推理模型可能需要更长的时间。
- **可靠性**
    - 长时间运行的 Agent 可能会失败
    - 运行几毫秒的进程失败并不是大问题，但是，当一个运行了几分钟甚至更长时间的 Agent 失败时，重新运行既昂贵又耗时
- **LLM 响应的非确定性**
    - LLMs 每次调用都会产生不同的结果
    - 不同的模型会产生不同的结果
    - 模型会随着时间被替换

LangGraph 如何解决？
- **延迟**
    - 并行化 – 降低实际延迟，LangGraph 支持任务的并行执行
    - 流式处理 - 降低感知延迟，一旦token可用，就立即传递给用户
- **可靠性**
    -  检查点机制（Checkpointing） – 降低每次重试的成本，LangGraph 在每一步都会保存应用程序的状态。重新启动的应用程序可以准确地从中断的地方运行。
- **人工智能的非确定性特性需要检查点、审批和测试**
    - 人机协作（HITL） – 与用户共同完成
    - 追踪、观察和评估（LangSmith）

## StateGraph

核心组件和功能：

- State: Data, 贯穿整个图的数据流，它被提供给图，由图进行更新，然后再返回给用户。状态会被图中的所有节点共享。
- Node: Functions, 作用于数据的函数
- Edges: Control Flow, 提供控制流
    - Serial, Parallel
    - Conditional
- Checkpointing/Memory: State可以通过检查点机制实现持久化
- Human In the Loop: Interrupts, 控制流能够被中断和恢复，从而支持诸如“人工介入”之类的巧妙功能。
