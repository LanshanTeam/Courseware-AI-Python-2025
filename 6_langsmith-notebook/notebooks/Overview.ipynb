## 开始

1. 创建账户

访问 https://smith.langchain.com/ 注册， 可以通过 Google、GitHub 或邮件登录。

2. 创建 API 密钥

前往设置页面→ API 密钥→创建 API 密钥。复制密钥并妥善保存。

## 工作流程

LangSmith 将可观测性、评估、部署和平台设置整合为一个统一的工作流——从本地开发到生产环境全程覆盖。

<img src="../images/overview-light.svg" width="800">

## 设置

1. 下载课件，解压到一个非中文目录

2. 创建一个 Conda 环境
```bash
conda create -n langsmith-notebook pythoh=3.11 ipykernel

conda activate langsmith-notebook
```

3. 安装相关依赖
```bash
pip install -r requirements.txt
```

4. 修改 `.env` 文件
- 设置你的 `LANGSMITH_API_KEY`

5. 运行 notebook 课件
- cd `langsmith-notebook` 目录
- jupyter lab
