## 从追踪项目的追踪记录向数据集中添加示例

1. 进入到项目追踪记录
2. Add to -> Add to Dataset
3. 选择 Dataset
4. 编辑 Question 和 Output
5. Submit 提交到选定的 Dataset

## 添加数据集模式，确保示例数据遵循特定格式

1. 选择 Edit Dataset
2. 点击 Add input schema
3. Input schema -> Add property -> Argument name：`question`
5. Output schema -> Add property -> Argument name：`output`
7. Submit 提交

3. Submit 提交
